```python
vowels = "аеёиоуыэюя"
students = input().split(",")
for student in students:
    words = student.split()
    if not words:
        continue
    count = 0
    for ch in student.lower():
        if ch in vowels:
            count += 1
    initials = "".join(w[0].upper() + "." for w in words)
    print(initials, count)
```
