```python
n = int(input())
runs = []
for _ in range(n):
    m = int(input())
    room = []
    for _ in range(m):
        total_str, name = input().split()
        cents = round(float(total_str) * 100)  
        room.append((-cents, name, total_str))
    runs.append(room)
while len(runs) > 1:
    new_runs = []
    for k in range(0, len(runs), 2):
        if k + 1 == len(runs):
            new_runs.append(runs[k])
            continue
        a, b = runs[k], runs[k + 1]
        merged = []
        i = j = 0
        while i < len(a) and j < len(b):
            if a[i] <= b[j]:
                merged.append(a[i])
                i += 1
            else:
                merged.append(b[j])
                j += 1
        while i < len(a):
            merged.append(a[i])
            i += 1
        while j < len(b):
            merged.append(b[j])
            j += 1
        new_runs.append(merged)
    runs = new_runs
result = runs[0]
out = [str(len(result))]
for rec in result:
    out.append(f"{float(rec[2]):.2f} {rec[1]}")
print("\n".join(out))
```
